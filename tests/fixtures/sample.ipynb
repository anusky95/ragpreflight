# RAG Quality Check Demo
This notebook shows how to use RAGCheck before embedding your documents.

```
from ragcheck import scan_document
report = scan_document("document.pdf")
print(report.score)
```

## Interpreting Results
Scores above 80 are ready for ingestion. Below 60 needs remediation.

```
for issue in report.issues:
    print(issue.severity.value, issue.message)
```
